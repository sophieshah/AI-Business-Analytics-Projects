```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns; sns.set()
import numpy as np

rng = np.random.RandomState(5933)
x = 10 * rng.rand(50)
y = 102 * x - 5 + rng.randn(50)
plt.scatter(x, y)

from sklearn.linear_model import LinearRegression
model = LinearRegression(fit_intercept=True)

model.fit(x[:, np.newaxis], y)

xfit = np.linspace(0, 10, 1000)
yfit = model.predict(xfit[:, np.newaxis])

plt.scatter(x, y)
plt.plot(xfit, yfit);


###HINT! You don't need to run this graph to answer... but you can! Look at the first few lines of code where X and Y are created.
## x = 10 * rng.rand(50)
## y = 102 * x - 5 + rng.randn(50)
### notice how y is generated from X? Which direction should that relationship go in? And how small or large is y in relation is x?
```
